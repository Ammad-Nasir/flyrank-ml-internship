```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
'''
I loaded the starter dataset and checked some basic information about it.
'''
import pandas as pd

# Load the FlyRank starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/Ammad-Nasir/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv")

# Basic dataset information
print("Dataset shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Show the first 5 rows
print(display(df.head()))
display(df.tail())

'''
What I found

The dataset has [30000] rows and [44] columns.

There are [30000] pages with a down trend.

The average number of impressions over the 90-day period is [YOUR NUMBER], while the average number of clicks is [YOUR NUMBER].

These numbers make this problem interesting because there are many pages with different levels of search performance.
 This gives me something to investigate when deciding which pages might be worth checking first.
 '''
```

```python
# Pages with at least 100 impressions but zero clicks
zero_click_pages = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print(
    "Pages with 100+ impressions but 0 clicks:",
    len(zero_click_pages)
)

print(
    "Percentage of all pages:",
    round(len(zero_click_pages) / len(df) * 100, 2),
    "%"
)
```
